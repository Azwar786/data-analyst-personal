4) From the DataFrame of Q1, select (a) only the Name and Salary columns, (b) rows where Age is greater than 30, (c) rows where City is 'Delhi' and Salary is above 40000.

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [25, 35, 30, 40, 28, 45],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Kolkata', 'Delhi', 'Delhi'],
    'Salary': [45000, 35000, 50000, 42000, 39000, 60000]
}
df = pd.DataFrame(data)

print("(a) Name and Salary columns:")
print(df[['Name', 'Salary']])

print("\n(b) Age greater than 30:")
print(df[df['Age'] > 30])

print("\n(c) City is Delhi and Salary above 40000:")
print(df[(df['City'] == 'Delhi') & (df['Salary'] > 40000)])
```

5) Use loc and iloc to (a) get the 2nd row, (b) get Name of the 4th employee, (c) get rows 1 to 3 with columns Name and City.

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [25, 30, 35, 28, 32, 40],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney', 'Berlin'],
    'Salary': [70000, 80000, 90000, 75000, 85000, 95000]
}
df = pd.DataFrame(data)

df['Bonus'] = df['Salary'] * 0.10
df['Total'] = df['Salary'] + df['Bonus']
df = df.rename(columns={'City': 'Location'})
df = df.drop(columns=['Bonus'])

print(df)

```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, np.nan, 35, 28, 32],
    'Salary': [70000, 80000, np.nan, 75000, 85000]
}
df = pd.DataFrame(data)

print("Original DataFrame with NaNs:")
print(df)

print("\nMissing values count per column:")
print(df.isnull().sum())

# Fill numeric NaNs with column mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

# Drop rows that still have missing values
df = df.dropna()

print("\nFinal DataFrame after filling and dropping:")
print(df)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
sorted_df = df.sort_values(by=['Salary', 'Age'], ascending=[False, True])
print(sorted_df)

top_3 = df.nlargest(3, 'Salary')
print(top_3)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
import pandas as pd

# 1. Create employees DataFrame
emp_data = {
    'EmpID': [1, 2, 3, 4],
    'Name': ['Ali', 'Sara', 'Ahmed', 'Fatima'],
    'DeptID': [101, 102, 101, 103]
}
employees = pd.DataFrame(emp_data)

# 2. Create departments DataFrame
dept_data = {
    'DeptID': [101, 102, 104],
    'DeptName': ['HR', 'IT', 'Finance']
}
departments = pd.DataFrame(dept_data)

# 3. Merge using inner join
inner_merged = pd.merge(employees, departments, on='DeptID', how='inner')
print("--- Inner Join ---")
print(inner_merged)

# 4. Merge using left join
left_merged = pd.merge(employees, departments, on='DeptID', how='left')
print("\n--- Left Join ---")
print(left_merged)

# 5. Concatenate two DataFrames of the same structure and remove duplicates
df1 = pd.DataFrame({'Name': ['Ali', 'Sara'], 'Salary': [50000, 60000]})
df2 = pd.DataFrame({'Name': ['Sara', 'Bilal'], 'Salary': [60000, 70000]})

concatenated = pd.concat([df1, df2]).drop_duplicates()
print("\n--- Concatenated & Deduplicated ---")
print(concatenated)
```
