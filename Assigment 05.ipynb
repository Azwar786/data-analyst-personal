QUESTION 1: Working with Lists
1.	Create a list called numbers containing at least 8 integers.
2.	Using a for loop, do the following:
o	Print all numbers in the list.
o	Print the sum of all numbers.
o	Print only the even numbers.
Requirements:
•	Use at least one loop.
•	Use conditional statements.
•	Display clear output.


```python
numbers = [10, 15, 20, 25, 30, 35, 40, 45]

print("All numbers:")
for num in numbers:
    print(num)

total = 0
for num in numbers:
    total = total + num

print("Sum of all numbers:", total)

print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

QUESTION 2: Dictionary Basics
1.	Create a dictionary called student with the following keys:
o	name
o	age
o	course
o	marks (must store a list of 3 marks)
2.	Using loops:
o	Print all keys and their values.
o	Calculate the average of the marks.
o	Display whether the student Passed (average ≥ 50) or Failed.
Requirements:
•	Dictionary must contain a list.
•	Use loops to calculate the average.


```python
student = {
    "name": "Arooma",
    "age": 17,
    "course": "Python",
    "marks": [70, 65, 80]
}

print("Student Details:")
for key, value in student.items():
    print(key, ":", value)

total = 0

for mark in student["marks"]:
    total = total + mark

average = total / len(student["marks"])

print("Average Marks:", average)

if average >= 50:
    print("Result: Passed")
else:
    print("Result: Failed")
```


QUESTION 3: List of Dictionaries
1.	Create a list called employees.
2.	Each employee must be stored as a dictionary containing:
o	name
o	department
o	salary
3.	Using loops:
o	Print details of all employees.
o	Find the employee with the highest salary.
o	Calculate the total salary expense.
Requirements:
•	Minimum 3 employees.
•	Use comparison logic.
•	Use loops to process data.


```python


# 1. Create a list called employees containing dictionaries (minimum 3 employees)
employees = [
    {"name": "Ali Khan", "department": "Data Science", "salary": 85000},
    {"name": "Sara Ahmed", "department": "Marketing", "salary": 65000},
    {"name": "Usman Tariq", "department": "Engineering", "salary": 95000},
    {"name": "Ayesha Malik", "department": "Human Resources", "salary": 60000}
]

# 2. Print details of all employees using a loop
print("--- All Employees Details ---")
for emp in employees:
    print(f"Name: {emp['name']} | Department: {emp['department']} | Salary: ${emp['salary']}")

print("\n" + "-"*30 + "\n")

# 3. Find the employee with the highest salary using comparison logic and a loop
highest_paid = employees[0]  # Assume the first employee has the highest salary initially

for emp in employees:
    if emp["salary"] > highest_paid["salary"]:
        highest_paid = emp

print("--- Employee with the Highest Salary ---")
print(f"Name: {highest_paid['name']}")
print(f"Department: {highest_paid['department']}")
print(f"Salary: ${highest_paid['salary']}")

print("\n" + "-"*30 + "\n")

# 4. Calculate the total salary expense using a loop
total_expense = 0

for emp in employees:
    total_expense += emp["salary"]

print("--- Total Salary Expense ---")
print(f"Total Expense: ${total_expense}")
```

QUESTION 4: While Loop with User Input
1.	Ask the user to enter numbers.
2.	Store the numbers in a list.
3.	Stop when the user enters -1.
4.	After the loop ends:
o	Print the complete list.
o	Print the largest number.
o	Print the smallest number.
Requirements:
•	Must use a while loop.
•	Must store input in a list.
•	Handle user input correctly.


```python


# Pre-defined list of numbers where -1 stops the sequence
simulated_inputs = [15, 42, 8, 99, 23, -1]
numbers = []

print("Simulating user input...")

# Process numbers using a loop
for val in simulated_inputs:
    print(f"Entered a number: {val}")
    if val == -1:
        break
    numbers.append(val)

# After the loop ends:
if len(numbers) > 0:
    print("\n--- Complete List ---")
    print("Numbers entered:", numbers)
    
    print("Largest number:", max(numbers))
    print("Smallest number:", min(numbers))
else:
    print("\nNo numbers were entered.")
```

QUESTION 5: Word Frequency Counter
1.	Ask the user to enter a sentence.
2.	Convert the sentence into a list of words.
3.	Use a dictionary to count how many times each word appears.
4.	Print the word frequency clearly.
Example Format (for understanding only):
apple: 3
banana: 2
orange: 1
Requirements:
•	Use string splitting.
•	Use a dictionary for counting.
•	Use loops for processing.


```python
# 1. Ask the user to enter a sentence
Sentence = input("Enter a sentence: ")

# 2. convert the sentence into a list of words using string splitting (.split())
words_list = Sentence.split()

# 3. Use a dictionary and loops to count how many times each word appears
word_count_dict = {}

for word in words_list:
    # Convert to lowercase to ensure accurate counting
    word = word.lower()
    
    if word in word_count_dict:
        word_count_dict[word] += 1
    else:
        word_count_dict[word] = 1

# 4. Print the word frequency clearly in the requested format
print("word frequencies:")
for word, count in word_count_dict.items():
    print(f"{word}: {count}", end=" ")
```


QUESTION 6: Nested Loops – Multiplication Table
Using nested loops, print a multiplication table from 1 to 5.
Expected Output Format:
1 2 3 4 5
2 4 6 8 10
3 6 9 12 15
4 8 12 16 20
5 10 15 20 25
Requirements:
•	Must use nested loops.
•	Proper formatting required.

General Requirements
•	Use:
o	At least 2 lists
o	At least 2 dictionaries
o	At least 3 loops
o	At least 1 nested loop
•	Include comments explaining your code.
•	Proper indentation is required.
•	Code must run without errors.
•	Submit your .py file before the due date


```python
# Print a multiplication table from 1 to 5 using nested loops

print("Multiplication Table from 1 to 5:")

# Outer loop for the first number (1 to 5)
for i in range(1, 6):
    # Inner loop for the second number (1 to 5)
    for j in range(1, 6):
        # Calculate product and print with proper spacing
        product = i * j
        print(f"{product:2}", end=" ")
    print() # New line after each row
```
